### A short notebook on the Nyquist sampeling theorem

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def y(t,f, phi = 0):
    return np.sin(2*np.pi*f*t + phi)

fm = 20

t = np.arange(0,1,1e-4)
ym = y(t,fm)
```

```python
# let us say we sample with fs = 19Hz
fs = 19
ts = np.arange(0,1,1/fs)
ys = y(ts,fm)

# let us say we sample with fs = 21Hz
fs2 = 21
ts2 = np.arange(0,1,1/fs2)
ys2 = y(ts2,fm)

# Now a sample with fs = 2*fm and sampeling smart
fc = 2*fm
t0 = 1/(4*fm)
tc = np.arange(t0,1,1/fc)
yc = y(tc,fm)
```

```python

plt.figure()
plt.plot(t,ym, label = f"f:{fm}")
plt.plot(ts,ys, '-o', label = f"f:{fs}", alpha = 0.5)
plt.plot(ts2,ys2, '-o', label = f"f:{fs2}", alpha = 0.5)
plt.plot(tc,yc, '-o', label = f"f:{fc}" , alpha = 0.5)

plt.legend()
plt.grid()
plt.show()


```
